A very critical issue I noticed is that the sensors are not the car themselves. The raw data is accurate, however not all the data is in the same manner. Such as the motor speed, which only updates every 2nd row. With foward fill, half the rows showed a change of 0 between them, further they're was also a bad reading which threw off acceleration calculations. Instead we found acceleration using a window of .5 seconds. Another instance of the sensors not being the car, is the motor RPM showing a higher speed than the IMU. The motor RPM tells you how fast the wheels should spin, accounting for perfect tire grip, however in reality the tire grip will compromise some of the motor RPM, causing a lower actual speed than the data provided speed.
