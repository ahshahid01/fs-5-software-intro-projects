# Step 1 - Forward fill and print the data
> Using parquet logic, forward fill and print all the values in a table format from the file.

Each sensor logs at its own rate, so most cells in the raw file are empty (null).
`fill_null(strategy="forward")` fills each empty cell with the last real reading above it in that column.

```python
import polars as pL

raw = pL.read_parquet("./software-data.parquet")
pd = raw.fill_null(strategy="forward")

print("rows, columns:", pd.shape)
print("empty cells before forward fill:", raw.null_count().sum_horizontal()[0])
print("empty cells after forward fill: ", pd.null_count().sum_horizontal()[0])
```
